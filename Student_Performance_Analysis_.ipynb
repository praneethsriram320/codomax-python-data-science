```python
import pandas as pd

df = pd.read_csv(
    '/content/student_data/student-mat.csv',
    sep=';'
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Display first 5 rows
print("First 5 rows:")
display(df.head())

# Check dataset shape
print("\nDataset Shape:")
print(df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='G3', bins=10, kde=True)

plt.title('Distribution of Final Student Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='studytime', y='G3')

plt.title('Study Time vs Final Student Grades')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='absences', y='G3')

plt.title('Absences vs Final Student Grades')
plt.xlabel('Number of Absences')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(df['G1'], color='blue', label='G1', kde=True)
sns.histplot(df['G2'], color='green', label='G2', kde=True)
sns.histplot(df['G3'], color='red', label='G3', kde=True)

plt.title('Comparison of Student Grades')
plt.xlabel('Grades')
plt.ylabel('Number of Students')

plt.legend()
plt.show()
```
